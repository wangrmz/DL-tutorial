```python
import torch
```

```python
x = torch.tensor(2.0, requires_grad=True)
y = x.detach()
print(x)
print(y)
'''
tensor(2., requires_grad=True)
tensor(2.)
'''
```

```python
print(x.requires_grad) #  True
print(y.requires_grad) # False
```

```python
print(id(x))
print(id(y))
```

```python
# 分别对x , y 进行后续计算
z1 = x ** 2
z2 = y ** 2
print(z1) # tensor(4., grad_fn=<PowBackward0>)
print(z2) # tensor(4.)
```

```python
# 获取梯度
z1.sum().backward()
```

```python
x = torch.ones(2,2, requires_grad=True)
print(x.grad)
y = x * x
# 分离y来返回一个新变量u
u = y.detach()
z = u * x
# 梯度不会向后流经u到x
z.sum().backward()
# 反向传播函数计算z=u*x关于x的偏导数时将u作为常数处理，而不是z=x*x*x关于x的偏导数
x.grad == u
# tensor([[True, True],
#       [True, True]])
```
