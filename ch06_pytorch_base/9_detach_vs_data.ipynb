```python
import  torch
```

```python
# 用两组数据进行对比
x1 = torch.tensor([1.0,2.0,3.0],requires_grad=True)
x2 = torch.tensor([1.0,2.0,3.0],requires_grad=True)
print(x1)
print(x2)
'''
tensor([1., 2., 3.], requires_grad=True)
tensor([1., 2., 3.], requires_grad=True)
'''
```

```python
y1 = x1.sigmoid()
y2 = x2.sigmoid()
print(y1)
print(y2)
'''
tensor([0.7311, 0.8808, 0.9526], grad_fn=<SigmoidBackward0>)
tensor([0.7311, 0.8808, 0.9526], grad_fn=<SigmoidBackward0>)
'''
```

```python
# 区别对待
# 都是用来底层用来存储数据的
z1 = y1.data # 只做读取，不做修改，修改后会影响梯度计算
z2 = y2.detach()
print(z1)
print(z2)
'''
tensor([0.7311, 0.8808, 0.9526])
tensor([0.7311, 0.8808, 0.9526])
'''
```

```python
print(z1.requires_grad)
print(z2.requires_grad)
```

```python
z1.zero_()
z2.zero_()
print(y1)
print(y2)

'''
tensor([0., 0., 0.], grad_fn=<SigmoidBackward0>)
tensor([0., 0., 0.], grad_fn=<SigmoidBackward0>)
'''
```
