# 张量运算函数

常见运算函数：
sum()求和
mean()求均值
max()/min()求最大/最小值及其索引
argmax()/argmin()求最大值/最小值的索引
std()求标准差
unique()去重
sort()排序

```python
import torch
```

```python
tensor1 = torch.randint(1,10,(3,2,4)).float()
print(tensor1)
```

```python
# sum 求和 这里是所有元素
print(tensor1.sum())
```

```python
# 按照维度求和，dim目前是三个维度
print(tensor1.sum(dim=0))
```

```python
# mean求均值
print(tensor1.mean())
```

```python
# max() 求最大值
print("求最大值")
print(tensor1.max())
print("按第2个维度求最大值与索引")
print(tensor1.max(dim=2))
print()
```

```python
# std() 求标准差
print("求标准差")
print(tensor1.std())
print()
```

```python
# argmin() 求最小值索引
print("求最小值索引")
print(tensor1.argmin())
print()
```

```python
# max() 求最大值
print("求最大值")
print(tensor1.max())
print("按第2个维度求最大值与索引")
print(tensor1.max(dim=2))
print()
```

```python
# unique() 去重
print("去重")
print(tensor1.unique())
print()
'''
去重
tensor([1., 2., 3., 4., 5., 7., 8., 9.])
'''
```

```python

# sort() 排序
print("排序")
print(tensor1.sort())
```
