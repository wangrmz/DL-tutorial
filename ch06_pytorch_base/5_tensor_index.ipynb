```python
import torch
```

```python
tensor1 = torch.randint(1, 10, (3, 5, 4))
print(tensor1)
```

```python
# 1.简单索引操作
print(tensor1[2, 1, 3])
```

```python
# 切片操作
print(tensor1[:, 3])
```

```python
# 范围索引
print(tensor1[1:])
```

```python
print(tensor1[-1])
```

```python
print(tensor1[-1:, 1:4, 0:3])
```

```python
# 1:4:2 最后一个2是步长（张量中的步长必须是正的）
print(tensor1[-1:, 1:4:2, 0:3])
```

```python
# 列表索引，注意对应关系
# 0 1
# 1 2
print(tensor1[[0, 1], [1, 2]])
```

```python
# 一对多的关系
print(tensor1[[[0], [1]], [1, 2]])
```

```python
# 布尔索引
print(tensor1[tensor1 > 5])
```

```python
# 选取符合条件的行，首元素大于 5
print(tensor1[:, :, 0] > 5)
```

```python
mask = tensor1[:, :, 0] > 5
print(tensor1[mask])
```

```python
# 选取符合条件的列，第二个元素大于 5
print(tensor1[:, 1, :] > 5)
```

```python
mask = tensor1[:, 1, :] > 5
# print(tensor1[mask]) # 不是全维度
# 处理原张量进行转置操作,只要实现后面两个翻转
# 如果直接转置的话，会处理成全翻转，原3 5 4 会翻转成4 5 3
tensor2 = tensor1.mT
```

```python
# 将mask应用到tensor2上，形状匹配前两个维度
print(tensor2[mask].mT)
```

```python
# 选取符合条件的矩阵：元素（1，2）> 5
mask = tensor1[:, 1, 2] > 5
print(mask)
```

```python
print(tensor1[mask])
```
