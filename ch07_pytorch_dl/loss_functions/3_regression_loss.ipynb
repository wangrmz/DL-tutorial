```python
import torch
import torch.nn as nn
```

```python
# 定义数据
input = torch.randn(3,5)
target = torch.randn(3,5)
print(input)
print(target)
```

```python
# L1 Loss对异常值鲁棒，但在0点处不可导。
# 平均绝对误差（Mean Absolute Error，MAE），也称L1 Loss：
```

```python
# 创建L1Loss对象
mae_loss= nn.L1Loss()
print(mae_loss(input,target)) # tensor(1.2578)
```

```python
#MSE ：L2Loss
mse_loss = nn.MSELoss()
print(mse_loss(input,target)) #tensor(2.2078)
```

```python
#Smoth ：L1
smooth_loss = nn.SmoothL1Loss()
print(smooth_loss(input,target)) #tensor(0.8308)
```
