```python
import torch
import torch.nn as nn
```

```python
# 1.目标值为顺序编码的标签
input = torch.randn(6, 8)
print(input)
# pred = torch.softmax(input, dim=1)
# print(pred)
```

```python
# 目标值，6个数据的类别标签（0 ～7）
target = torch.tensor([1, 0, 7, 5, 2, 3])  # 6个标签，对应6个样本
print(target)
```

注意：调用torch.nn.CrossEntropyLoss相当于调用了torch.nn.LogSoftmax之后再调用torch.nn.NLLLoss。即使用CrossEntropyLoss时上一层的输出不需要Softmax激活函数，因为该损失函数内会自动处理

CrossEntropyLoss = LogSoftmax + NLLLoss

内部流程:
input (logits)
  ↓
LogSoftmax  ← 自动执行
  ↓
NLLLoss (负对数似然损失)
  ↓
loss value



```python
# 定义损失函数
loss = nn.CrossEntropyLoss()
print(loss(input, target))
```

```python
# 2.目标值为一组概率（独热编码）
input = torch.randn(6, 8)
print( input)
```

```python
target = torch.randn(6,8).softmax(dim=1) # 按照行进行处理
print(target)
```

```python
loss = nn.CrossEntropyLoss()
print(loss(input, target))
```
